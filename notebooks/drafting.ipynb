# Part 1: PickNet: predicting (and simulating) human draft picks

The wheel model asks what the *other* seven drafters will take. This notebook models the drafter's own choice:
given the pack and everything seen so far, **which card does the human take?** A pick is a discrete choice
among exactly the cards in the pack, so every model outputs a softmax over the pack (a conditional logit, as in
DraftFM). The headline measures are **top-1 agreement** (the model's favourite card is the one the human took) and
**log loss per pick** (nats), on the same chronological test drafts (last 15%) as the wheel and deck models.

Models (`src/openlane/picknet.py`, trained by `scripts/train_picknet.py`):

| model | what it sees |
|---|---|
| `rating` | one learned number per card: a pick-order bot |
| `pool` | rating + a rank-d bilinear term between the candidate and the mean of the pool so far ("stay in your colours") |
| `picknet_nohist` | transformer: candidates attend to each other (which cards compete), state = pool bag + position + context |
| `picknet` | + a causal transformer over the whole draft so far: every pack seen, every pick, what the other seven took from packs that came back; candidates cross-attend to it |
| `picknet_big` | the same, d=192, 4 history + 3 pack layers |
| `multiset_zeroshot` | PickNet with **no card ids**: a card is only its features (colour, rarity, mana value, types, Scryfall rules features, a frozen rules-text embedding). Trained on 7 other sets, scored on MSH, which it never saw (DraftFM's "day zero" setup) |
| `multiset_plus_target` | the same features-only model, trained on the 7 sets + MSH's own training drafts |

Every model also gets per-draft context (rank, 17Lands skill buckets, days into the format), except `rating`.

## 1. Results on MSH

10,292 test drafts, 432k picks. The last pick of each pack has one card left, so it is always "right"; DraftFM
reports agreement both with and without these forced picks, and so does this table. Uniform random would score
about 1/n per pick: 22% overall, 18% without forced picks.

````python
def nonforced(m):
    z = np.load(P / f"test_{m}.npz") if (P / f"test_{m}.npz").exists() else None
    if z is None:  # every test draft has every pick, so the per-pick curves give the same means
        t1, nl = (np.array(res[m][k], float) for k in ("top1_by_pick", "nll_by_pick"))
        N = len(t1) // 3
        keep = np.ones(len(t1), bool)
        keep[N - 1::N] = False
        return t1[keep].mean(), nl[keep].mean()
    N = z["top1"].shape[1] // 3
    keep = np.ones(z["top1"].shape[1], bool)
    keep[N - 1::N] = False
    return z["top1"][:, keep].mean(), z["nll"][:, keep].mean()

rows = []
for m, r in res.items():
    nf1, nfl = nonforced(m)
    rows.append({"model": m, "params": r.get("params"), "top-1": r["top1"], "log loss": r["nll"],
                 "top-1 non-forced": nf1, "log loss non-forced": nfl,
                 "P1 / P2 / P3 top-1": f"{r['top1_p1']:.3f} / {r['top1_p2']:.3f} / {r['top1_p3']:.3f}"})
rows.append({"model": "DraftFM, MSH zero-shot (paper, d=128)", "params": 1_638_000, "top-1": 0.5671,
             "log loss": 1.1352, "top-1 non-forced": 0.5338, "log loss non-forced": None, "P1 / P2 / P3 top-1": ""})
tbl = pl.DataFrame(rows)
tbl.with_columns(pl.col("top-1", "log loss", "top-1 non-forced", "log loss non-forced").round(4))
````

DraftFM's MSH numbers are over a different (earlier) slice of MSH drafts and pool Premier Draft drafts from the

## 6. Simulating whole tables

`scripts/simulate_draft.py` seats eight copies of a model at a table and drafts. The packs are real: each table
takes 8 random test drafts, and seat s opens the three packs that drafter opened, with that drafter's rank, skill
and date as context. Packs pass left, right, left. Each bot **samples** its pick from the model's softmax, so this
simulates human choices rather than playing the best card. 1,000 tables = 8,000 simulated drafts.

The simulated drafts are scored with the same statistics as the real test drafts (one 17Lands seat at a real
table):

* ATA and ALSA per card (cards taken 200+ times in the real drafts)
* the per-card wheel rate at picks 1-6. (The rate *by pick* is fixed by pack size: 6 of the 13 cards passed at
  pick 1 always come back, so it can't tell models apart.)
* **colour commitment**: of the drafter's last 14 picks, the share in their two main colours
* the share of drafters in each two-colour pair

````python
rows = []
for m in ["rating", "pool_d32", "picknet", "picknet_T0.7", "picknet_big"]:
    f = P / f"sim_{m}.json"
    if f.exists():
        r = json.loads(f.read_text())
        rows.append({"bots": m, "ATA r": r["ata_corr"], "ATA MAE": r["ata_mae"], "ALSA r": r["alsa_corr"],
                     "wheel r (per card)": r["wheel_card_corr"], "wheel MAE": r["wheel_card_mae"],
                     "pack-3 picks in top-2 colours": r["commit_sim"], "real": r["commit_real"],
                     "colour-pair TV distance": r["pair_share_tv"], "P1P1 r": r["p1p1_corr"]})
sim = pl.DataFrame(rows).with_columns(pl.exclude("bots").round(3))
sim
````

`picknet_T0.7` samples at temperature 0.7: sharper picks over-commit (0.828 of pack-3 picks in the top two
colours, vs 0.815 for humans) and still under-draft WU, so temperature 1 is the calibrated setting.

**The simulator as a laboratory.** `scripts/force_vs_follow.py` uses these tables to ask a strategy question:
seats 1–7 are human-like PickNet bots and seat 0 follows, forces a pair, or locks in early signals. Forcing a
random pair ties drafting like a human, forcing the strongest pair wins, and pure signal-chasing loses, on MSH
and FDN. See `drafting.ipynb` part 2 §4.

## 7. Takeaways

1. **PickNet predicts 68.2% of MSH test picks (big: 68.6%; features-only, pretrained on 7 other sets: 68.6%), against 49.7% for a pick-order bot and 64.4% for the
   bilinear pool model.** Log loss falls from 1.324 to 0.826 (big: 0.815) nats per pick. That is close to the
   ~70-71% reported for in-set models on other, larger sets (statistical-drafting, puder). Without forced last
   picks: 65.8% (big 66.2%).
2. **Where the gain comes from:** the pool is most of it (+14.7 points: stay in your colours), then the pack
   (+2.9: candidates attend to each other, so the model knows which cards compete), then the draft history
   (+0.9 points and −0.029 nats; the gain grows to +1.3 in pack 3). So the history of what was passed and what
   came back does help predict human picks, against puder's "+0.57% from pack contents".
3. **Better drafters are more predictable**, by about 4 points from Bronze to Mythic and across the 17Lands
   win-rate buckets, while the rating bot's agreement barely moves. Strong players converge on the same
   *contextual* pick, not just the same card ratings.
4. **Every set looks alike for the cheap models:** the pool model adds 15-22 points over a pick order on all 14
   sets, including 13- and 15-card packs.
5. **Day zero is hard.** With no card ids and 7 other sets (133k drafts), the features-only model gets 50.8% on
   MSH (non-forced 47.0%) against 68.6% on held-out drafts of the sets it trained on, and DraftFM's 56.7% with
   29 sets and 149M picks. The gap is largest at the first picks of each pack, where only card quality matters:
   judging power from rules text is the part that needs more sets.
   **Trained on those 7 sets plus MSH's own training drafts, the same features-only model is the best pick model
   in this notebook** (NLL 0.810, top-1 68.6%), ahead of PickNet with per-card ids (0.826) and level with the
   3x bigger id model (0.815). Shared card features plus other sets' drafts beat learning each card from scratch.
6. **As a simulator, PickNet drafts like people, and the cheap models don't.** Rating bots match per-card ATA and
   wheel rates well (those are marginal statistics, which a per-card rating fits directly) but only 56% of their
   pack-3 picks are in their two main colours (humans: 81.5%), and their colour pairs come out nearly uniform.
   PickNet bots commit like humans through every pack (0.80), pick P1P1 like humans (r = 0.97) and get the
   colour-pair mix closest (TV 0.066, pool 0.12, rating 0.15). All bots under-draft the format's most popular
   pair (WU: 23% of humans, 17% of PickNet bots): humans converge on the best archetype harder than a model that
   only imitates individual picks.
7. **Next:** temperature < 1 and a skill setting for the bots (sample as a Mythic drafter); use the simulator to
   produce wheel labels for the wheel model at all 8 seats; more sets (and the 13/15-card sets) for day zero.

# Part 2: Does the open lane pay?

The repo's namesake question: are table signals real, do drafters act on them, and does being in the open
lane win games? Then a counterfactual in the table simulator: is it better to follow signals or to force a
colour pair? Everything here reads the JSON results, so it runs in seconds. To regenerate:

```
python scripts/open_lane.py MSH [SET ...]       # ~1 min per set
python scripts/force_vs_follow.py MSH 2000      # ~10 min; needs results/MSH.PremierDraft/picknet/models/picknet.pt
```

**Openness** of colour *j* in pack *k* is how many good cards of that colour reached the drafter later than they
usually last:

  flow[d, k, j] = Σ over picks q = 1…7, over cards c in the pack shown at q, of colour_share[c, j] × max(0, q − ALSA[c])

Picks 8+ are excluded because those packs have gone round the whole table. It is z-scored per (pack, colour)
within each draft week, so set-wide shifts in colour popularity cancel. The drafter's own picks never enter it:
it is fixed by the opened packs and the upstream drafters. A drafter's **main colours** are the top two by
playability-weighted picks. Packs pass left in P1 and P3 and right in P2, so **P1 and P3 share the upstream
neighbour**.

````python
import json
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import polars as pl

ROOT = Path.cwd() if (Path.cwd() / "src").exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))

# reference categorical palette, fixed order (same as results.ipynb / wheelnet.ipynb)
BLUE, ORANGE, AQUA, YELLOW, MAGENTA, VIOLET = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#e87ba4", "#4a3aa7"
GRAY, INK, MUTED, SURFACE, GRID = "#a3a29c", "#0b0b0b", "#52514e", "#fcfcfb", "#e6e5e0"
plt.rcParams.update({
    "figure.dpi": 110, "axes.spines.top": False, "axes.spines.right": False,
    "axes.edgecolor": MUTED, "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.8, "axes.axisbelow": True,
    "lines.linewidth": 2, "font.size": 10, "legend.frameon": False,
})

OL = {f.parent.name.split(".")[0]: json.load(open(f))
      for f in sorted((ROOT / "results").glob("*.PremierDraft/open_lane.json"))}
print(f"{len(OL)} sets: {', '.join(OL)}")
````

## 1. Are signals real?

If openness reflects the neighbour's colours rather than noise from the packs, a colour that flowed in P1 should
flow again in P3 (same neighbour) more than in P2 (the other neighbour).

````python
pers = pl.DataFrame([{"set": s, "P1–P3 (same neighbour)": r["persistence"]["P1-P3 (same neighbour)"],
                      "P1–P2": r["persistence"]["P1-P2"], "P2–P3": r["persistence"]["P2-P3"]} for s, r in OL.items()]
                    ).sort("P1–P3 (same neighbour)")
fig, ax = plt.subplots(figsize=(7.5, 0.36 * pers.height + 1.2))
yy = np.arange(pers.height)
ax.hlines(yy, pers["P1–P2"], pers["P1–P3 (same neighbour)"], color=GRID, lw=3)
ax.scatter(pers["P1–P2"], yy, color=GRAY, s=48, zorder=3, label="P1 vs P2 (other neighbour)")
ax.scatter(pers["P1–P3 (same neighbour)"], yy, color=BLUE, s=48, zorder=3, label="P1 vs P3 (same neighbour)")
ax.set_yticks(yy, pers["set"]); ax.grid(axis="y", visible=False); ax.set_xlim(0, 0.3)
ax.set_xlabel("correlation of a colour's openness between rounds (mean over 5 colours)")
ax.legend(loc="upper left", bbox_to_anchor=(0, -0.13), ncol=2); ax.set_title("P1 signals predict P3, through the same neighbour", loc="left")
plt.show()
pers.sort("set")
````

## 2. Do drafters follow them?

Logit of "colour *j* ends up a main colour" on its P1 openness (with colour fixed effects). The slope is how much
one sd of openness moves the choice, by rank and by the 17Lands win-rate bucket.

## 3. Does it pay?

What this shows (all 15 sets):
1. **Signals are real and travel through your neighbour.** A colour's P1 openness correlates 0.22 with its P3
   openness and 0.09 with P2.
2. **Drafters follow, and better drafters follow more.** +1 sd of P1 openness raises the log-odds that a colour
   ends up in the deck by 0.36, rising from bronze to mythic and with the skill bucket on every set. Still, only
   ~17% end in the most open P1 pair (10% by chance).
3. **Being in the open lane wins games.** +0.055 log-odds per game per sd with skill controlled, on 15/15 sets;
   bottom to top quintile is about +3 points of game win rate.
4. **But the payoff is supply, not the read itself.** P2 openness, which you cannot read before committing, pays
   as much as P1, and openness in the colours you *didn't* take has no effect (≈ 0, negative on only 9/15 sets).
   Being where the cards are wins; this data can't separate the decision from the luck of the seat. Hence the
   counterfactual below.

## 4. Counterfactual: follow or force? (`scripts/force_vs_follow.py`)

The PickNet table simulator (`drafting.ipynb`): 2,000 tables of real opened packs, seats 1–7 always the
human-like PickNet bot, seat 0 plays a strategy. Every strategy runs on the same tables and random stream, so
differences are paired.

* **follow**: seat 0 is the human-like bot too (it reads signals as people do)
* **force**: sample from the same model, but only among cards that fit a fixed pair (colourless always fits);
  table *g* forces pair *g* mod 10
* **read 5, then lock**: free picks for P1 picks 1–5, then lock the pair that looks most open by the openness
  measure above, ignoring the cards already taken

Pools are scored as decks: for each pair, the 23 best fitting cards by training-game GIH WR (missing slots get a
10th-percentile card). The deck score is the best pair's mean.

````python
FV = {f.parent.name.split(".")[0]: json.load(open(f)) for f in sorted((ROOT / "results").glob("*.PremierDraft/force_vs_follow.json"))}
rows = []
for s, r in FV.items():
    bp = r["force"]["by_pair"]; best = max(bp, key=bp.get); f = r["follow"]["seat0"]
    rows += [{"set": s, "strategy": "follow (human-like)", "deck score": 100 * f, "vs follow": 0.0, "± se": None},
             {"set": s, "strategy": "force a random pair", "deck score": 100 * r["force"]["seat0"],
              "vs follow": 100 * r["force"]["vs_follow"], "± se": 100 * r["force"]["vs_follow_se"]},
             {"set": s, "strategy": f"force the best pair ({best})", "deck score": 100 * bp[best],
              "vs follow": 100 * (bp[best] - f), "± se": None},
             {"set": s, "strategy": "read 5, then lock", "deck score": 100 * r["force_late"]["seat0"],
              "vs follow": 100 * r["force_late"]["vs_follow"], "± se": 100 * r["force_late"]["vs_follow_se"]}]
pl.DataFrame(rows)
````

````python
fig, axes = plt.subplots(1, len(FV), figsize=(5.2 * len(FV), 3.6), sharey=True, squeeze=False)
for ax, (s, r) in zip(axes[0], FV.items()):
    bp = r["force"]["by_pair"]; order = sorted(bp, key=lambda k: -bp[k])
    f = 100 * r["follow"]["seat0"]
    v = np.array([100 * bp[k] for k in order])
    ax.bar(range(10), v - f, color=[BLUE if d > 0 else ORANGE for d in v - f], width=0.7)
    ax.axhline(0, color=MUTED, lw=1)
    ax.axhline(100 * r["force_late"]["vs_follow"], color=GRAY, ls="--", lw=1.2)
    ax.text(-0.5, 100 * r["force_late"]["vs_follow"] - 0.04, "read 5, then lock", fontsize=8, color=MUTED, va="top")
    ax.set_xticks(range(10), order); ax.set_xlim(-0.6, 10.6); ax.grid(axis="x", visible=False)
    ax.set_title(f"{s}: forcing each pair vs following", loc="left", fontsize=10)
axes[0][0].set_ylabel("deck score vs following\n(GIH WR points, 23-card deck)")
plt.show()
````

What this shows (MSH and FDN):
1. **Forcing a random pair costs nothing** in simulation (±0.03 points): committing at P1P1 ties drafting like a
   human.
2. **Which pair matters much more than when you commit.** Forcing the strongest pair beats following (+1.3 MSH,
   +0.5 FDN), and 2.9 / 0.9 points separate the best and worst pair to force.
3. **Pure signal-chasing loses.** Locking whatever looks open after five picks, ignoring the cards already taken,
   is worst on both sets (−0.4 / −0.3). Human-style drafting balances card quality, pool and signals.

Caveats: PickNet bots under-draft MSH's best pair (WU is 17% of bot decks vs 23% of real drafters), so a WU forcer
faces less competition than at a real table. The deck score ignores curve and synergy. And every seat imitates
17Lands users, not strong drafters. A 7-forcers table, or bots that also force, would test how forcing breaks
down when others do it too.
